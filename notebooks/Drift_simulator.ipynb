```python
import pandas as pd

df = pd.read_csv("../data/raw/loan_default.csv")



split = int(len(df) * 0.7)
reference = df.iloc[:split].copy()
current = df.iloc[split:].copy()

# Create drift Simulate crisis attack
current["Income"] *= 0.7
current["CreditScore"] -= 80
current["LoanAmount"] *= 1.2
current["DTIRatio"] *= 1.5


df_drifted = pd.concat([reference, current], ignore_index=True)


df_drifted.to_csv("loan_default_drifted.csv", index=False)

print("DONE: loan_default_drifted.csv created.")
```
